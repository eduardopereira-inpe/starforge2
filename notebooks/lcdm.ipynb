```python
## Importando bibliotecas necessárias
import numpy as np
import matplotlib.pyplot as plt

from starforge.cosmology import LCDMModel  
from starforge.cosmology import CosmologicalParameters
```

```python
## Definindo parâmetros cosmológicos (exemplo Planck 2018)
params = CosmologicalParameters(
    omegam=0.24,
    omegab=0.04,
    omegal=0.73,
    h=0.7   
)

lcdm = LCDMModel(params)
```

```python
## Exemplo 1: Evolução do parâmetro de Hubble H(z)
z = np.linspace(0, 10, 200)
Hz = lcdm.H(z)

plt.figure(figsize=(7,5))
plt.plot(z, Hz, label="H(z)")
plt.xlabel("Redshift z")
plt.ylabel("H(z) [km/s/Mpc]")
plt.title("Hubble Parameter Evolution")
plt.legend()
plt.grid()
plt.show()
```

```python
## Exemplo 2: Idade do Universo como função do redshift
z = np.linspace(0, 10, 300)
ages = [lcdm.age(zi) for zi in z]

plt.figure(figsize=(7,5))
plt.plot(z, ages, color="darkorange", label="Age(z)")
plt.xlabel("Redshift z")
plt.ylabel("Age of Universe [years]")
plt.title("Cosmic Age vs Redshift")
plt.legend()
plt.grid()
plt.show()
```

```python
## Exemplo 3: Densidade de matéria escura e bariônica com o redshift
z = np.linspace(0, 5, 200)
rodm = [lcdm.rodm(zi) for zi in z]
robr = [lcdm.robr(zi) for zi in z]

plt.figure(figsize=(7,5))
plt.plot(z, rodm, label="Dark Matter Density ρ_dm(z)")
plt.plot(z, robr, label="Baryonic Density ρ_b(z)")
plt.yscale("log")
plt.xlabel("Redshift z")
plt.ylabel("Density [arbitrary units]")
plt.title("Matter Densities vs Redshift")
plt.legend()
plt.grid()
plt.show()
```

```python
## Exemplo 4: Elemento de volume comóvel dV/dz
z = np.linspace(0.01, 5, 200)   # começa em 0.01 para evitar singularidade
dVdz = [lcdm.dV_dz(zi) for zi in z]

plt.figure(figsize=(7,5))
plt.plot(z, dVdz, color="green", label="dV/dz")
plt.xlabel("Redshift z")
plt.ylabel("Comoving Volume Element dV/dz [Mpc³/sr]")
plt.title("Comoving Volume Element vs Redshift")
plt.legend()
plt.grid()
plt.show()
```

```python
## Exemplo 4: Elemento de volume comóvel dV/dz
z = np.linspace(0.01, 5, 200)   # começa em 0.01 para evitar singularidade
dVdz = [lcdm.dt_dz(zi) for zi in z]

plt.figure(figsize=(7,5))
plt.plot(z, dVdz, color="green", label="dV/dz")
plt.xlabel("Redshift z")
plt.ylabel("Time Variation")
plt.legend()
plt.grid()
plt.show()
```

```python
## Exemplo 4: Elemento de volume comóvel dV/dz
z = np.linspace(0.01, 10, 200)   # começa em 0.01 para evitar singularidade
dVdz = [lcdm.rodm(zi) for zi in z]

plt.figure(figsize=(7,5))
plt.plot(z, dVdz, color="green", label=r"$\rho_{dm}$")
plt.xlabel("Redshift z")
plt.ylabel(r"$\rho_{dm}$")
plt.legend()
plt.grid()
plt.show()
```
